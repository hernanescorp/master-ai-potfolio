## Data Types
Numpy provides an alternative implementation for numerical arrays, improving the performance of data driven computation to standard Python built-in lists.
### Python Integers vs C integers
* Python int are complex objects written in C
    * Dynamically typed language
    * Almost infinite integer arithmetic precision
```c
struct _longobject {
    long ob_refcnt;         # reference count
    PyTypeObject *ob_type;  # type of the variable
    size_t ob_size;         # size of the following data members
    long ob_digit[1];       # integer value encoded into a long array
};
``` 
* Clanguage integers are simple references to a position in memory whose bytes encode an integer value.

## Python lists vs Numpy Arrays
* Python lists are complex objects, lists are heterogeneous, different object types and sizes, contain an array with references to each object, contain a pointer to a block of pointers, each of which points to a python object.
* Standart Numpy arrays are homogeneous.
    * Contains a single pointer to one contiguos block of data.

## Parallel computing & Random Number Generators (RNG)
When you run a program in parallel across multiple workers, usually want each worker to generate its own random numbers
* Give every worker the same seed --> the exact same sequence of numbers
* Give every worker different seeds --> sequence overlap & stadistical correlations

The solution is to give an independient RNG to each worker
1. parent_RNG = np.random.default_rng(seed) --> returns a parent RNG object.
2. children_RNG = parent_RNG.spawn(size) --> creates n mathematically independent RNG 
3. Use children_RNG[i]

You get
* Statistical independence
* Total reproducibility

### Array Dimensions
Numpy arrays can have any number of dimensions:
* 0-dimensional arrays --> **scalars** or **rank-0 tensor**
* 1-dimensional arrays --> vectors or rank-1 tensor
* 2-dimensional arrays --> Matrixs or rank-2 tensor
* 3-dimensional arrays --> tensors or rank-3 tensor
* N - dimensional arrays --> tensors or rank N tensor

### Array indexing
One dimensional arrays are indexed as Python lists.

````python
x = np.arange(100, 110)
print(x)

print(x[3])

#negative indexes are valid as with python lists:
print(x[-1])
````

Multi-dimensional arrays are indexed with a comma separated tuple of indices

````python
x = np.random.rand(3,4)
print(x)
print(x[0,3], x[-1,0])
````

Values can be modified using index

````python
x = np.arange(10)
print(x)
x[-1] = 20
print(x)
````

Numpy arrays have fixed types. If u set a float value to an int array, the value is truncated.

````python
print(x)
x[-1] = 1.5 
print(x)
````

## Data Types
Numpy provides an alternative implementation for numerical arrays, improving the performance of data driven computation to standard Python built-in lists.
### Python Integers vs C integers
* Python int are complex objects written in C
    * Dynamically typed language
    * Almost infinite integer arithmetic precision
```c
struct _longobject {
    long ob_refcnt;         # reference count
    PyTypeObject *ob_type;  # type of the variable
    size_t ob_size;         # size of the following data members
    long ob_digit[1];       # integer value encoded into a long array
};
``` 
* Clanguage integers are simple references to a position in memory whose bytes encode an integer value.

## Python lists vs Numpy Arrays
* Python lists are complex objects, lists are heterogeneous, different object types and sizes, contain an array with references to each object, contain a pointer to a block of pointers, each of which points to a python object.
* Standart Numpy arrays are homogeneous.
    * Contains a single pointer to one contiguos block of data.

## Parallel computing & Random Number Generators (RNG)
When you run a program in parallel across multiple workers, usually want each worker to generate its own random numbers
* Give every worker the same seed --> the exact same sequence of numbers
* Give every worker different seeds --> sequence overlap & stadistical correlations

The solution is to give an independient RNG to each worker
1. parent_RNG = np.random.default_rng(seed) --> returns a parent RNG object.
2. children_RNG = parent_RNG.spawn(size) --> creates n mathematically independent RNG 
3. Use children_RNG[i]

You get
* Statistical independence
* Total reproducibility

### Array Slicing
x[start:stop:step]

````python
x = np.arange(100,110)
print(f'{x[:5]=}')
print(f'{x[5:]=}')
print(f'{x[1:3]=}')
print(f'{x[::2]=}')
print(f'{x[::-2]=}')
print(f'{x[5::-1]=}')
````

````python
x = np.random.randint(1,10,(3,5))
print(f'{x=}')
print(f'{x[1,:]=}')
print(f'{x[:,1]=}')
print(f'{x[:2,3:]}')
````

Unlike python list slices, array slices return views rather than copies.
Explicit copies of arrays or subarrays slices can be created: **x.copy()**

````python
x = np.random.rand(3,5)
print(f'{x=}')
y = x[1,::2].copy()
print(f'{y=}')
````

### Reshaping of arrays

np.newaxis, when used in Array slicing adds a new dimension does a reshape

````python
from numpy import newaxis
a = np.arange(4)
b = a[newaxis,:] #row vector a.reshape(1,4)
c = a[:,newaxis] #column vector a.reshape((4,1))

print(f'{a=}\n{b=}\n{c=}')
````

### Array concatenation and splitting

* Concatenation : Combine multiple arrays into one, **np.concatenate, np.vstack, np.hstack**
* Splitting, split a single array into multiple arrays, **np.split, np.split, np.hsplit**



````python
a = np.arange(0,4)
b = np.arange(4,8)
c = np.arange(8, 12)
d = np.concatenate((a,b,c))
print(f'{a=}\n{b=}\n{c=}\n{d=}')
````

````python
a2, b2, c2 = a[np.newaxis, :], b[np.newaxis, :], c[np.newaxis, :]

d2 = np.concatenate((a2, b2, c2), axis=0)

print(f'{a2=}\n{b2=}\n{c2=}\n{d2=}')
````

````python
np.concatenate((a2,b2,c2), axis=1)
````

````python
a = np.arange(0,8).reshape((2,4))
b = np.arange(8,16).reshape((2,4))
c = np.hstack((a,b))
d = np.concatenate((a,b), axis = 1)

e = np.vstack((a,b))
f = np.concatenate((a,b), axis = 0)

print(f'{a=}\n{b=}\n{c=}\n{d=}\n{e=}\n{f=}')
````

````python
a = np.arange(42).reshape((6,7))
upper, center, lower = np.split(a,3)
print(f'{a=}\n{upper=}\n{center=}\n{lower=}')
````

````python
a = np.arange(36).reshape((6,6))
upper, center, lower = np.split(a, (1,3))
print(f'{a=}\n{upper=}\n{center=}\n{lower=}')
````

````python
a = np.arange(16).reshape((4,4))
b, c = np.vsplit(a,2)
d, e = np.split(a,2, axis=0)
print(f'{a=}\n{b=}\n{c=}\n{d=}\n{e=}')
````

````python
a = np.arange(16).reshape((4,4))
b, c = np.hsplit(a,2) 
d, e = np.split(a,2, axis=1)
print(f'{a=}\n{b=}\n{c=}\n{d=}\n{e=}')
````

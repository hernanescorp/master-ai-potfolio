### Python Integers vs C integers
* Python int are complex objects written in C
    * Dynamically typed language
    * Almost infinite integer arithmetic precision

## Python lists vs Numpy Arrays
* Python lists are complex objects, lists are heterogeneous, different object types and sizes, contain an array with references to each object, contain a pointer to a block of pointers, each of which points to a python object.
* Standart Numpy arrays are homogeneous.
    * Contains a single pointer to one contiguos block of data.

## Parallel computing & Random Number Generators (RNG)
When you run a program in parallel across multiple workers, usually want each worker to generate its own random numbers
* Give every worker the same seed --> the exact same sequence of numbers
* Give every worker different seeds --> sequence overlap & stadistical correlations

The solution is to give an independient RNG to each worker
1. parent_RNG = np.random.default_rng(seed) --> returns a parent RNG object.
2. children_RNG = parent_RNG.spawn(size) --> creates n mathematically independent RNG 
3. Use children_RNG[i]

You get
* Statistical independence
* Total reproducibility

## UFuncs - Universal Functions

Function designed to perform element wise operations on arrays. Main features:
* Speed and Efficency, instead of looping over arrays to perform teh same computation on each element, use single highly ooptimized functions.
* Broadcasting: Support for different but compatible shape arrays
* Type casting: Support for different data types

### Arithmetic operations

````python
x = np.arange(5)

print("x =", x)
print("x + 5 =", x + 5)
print("x - 5 =", x - 5)
print("-x", -x)
print("x * 2=", x*2 )
print("x / 2=", x/2)
print("x // 2 =", x // 2)
print(" x % 2 =", x % 2)
print("x ** 2 =", x ** 2)
````

## Broadcasting
The ability to apply binary ufuncs (addition, subtraction, multiplication) on arrays of diffferent sizes 

````python
a = np.arange(10)
b = a[::-1]

print(f'{a=}')
print(f'{b=}')
print(f'{a+b =}')
print(f'{a + 3=}') # broadcasting
````

### Broadcasting Rules

Given two arrays `x` and `y` with dimensions $(d_1^x,\dots,d_n^x)$ and $(d_1^y,\dots,d_m^y)$
1. if $n \ne m$, the shape of the one with fewer dimensions is padded with *ones* on its left side.
   * $n < m \to (1_1,\dots,1_{m-n},d_1^x,\dots,d_n^x)$
   * $n > m \to (1_1,\dots,1_{n-m},d_1^y,\dots,d_m^y)$
   * After the padding, shapes of `x` and `y` are  $(d_1^x,\dots,d_k^x)$ and $(d_1^y,\dots,d_k^y)$ , where $k = \max(n,m)$
1. If the shape of the two arrays does not match in any dimension, the array with shape equal to 1 in that dimension is stretched (replicated) to match the other shape.
   * After the stretching, $\forall i \;\; d_i^x = 1 \iff d_i^y = 1$
1. If in any dimension the sizes disagree ($\exists i \; where \; d_i^x \ne d_i^y$), an error is raised
   * If no error is raised, shape of `x` and `y` is $(d_1,\dots,d_k)$
3. All dimensions match, and shape of result is $(d_1,\dots,d_k)$

## Fancy Indexing

* Use arrays of indices to access multiple array elemnts
* Returns a copy of the data (vs slicing view)

````python
import numpy as np
a = np.random.randint(100, size=(10,))
ii1 = [1,8,3,5]
ii2 = np.array([1,8,3,5])

print(f'{a=}')
print(f'{a[ii1]=}')
print(f'{a[ii2]=}')
````

Fancy indexing can be used with multi-dimensional arrays
* Use arrays of indices to access multiple array elemnts
* Sequential selection: returns 1-dimensional array

````python
a = np.arange(15).reshape((3,5))
ii = [1,0,2,2]
jj = [1,4,3,2]

print(f'{a =}')
print(f'{a[ii,jj]= }')
````

````python
a = np.arange(16).reshape((4,4))
ii = [1,3,2]

print(f'{a=}')
print(f'{a[1, ii] =}')
print(f'{a[ii, 2] =}')
````

````python
a = np.arange(15).reshape((3,5))

ii = [2,0]
jj = [1,4,3]

print(f'{a=}')
print(f'{a[ii] =}')
print(f'{a[ii, :]=}')
print(f'{a[:,jj]}')
````

#### Secuence Unpacking
- Python Allows multiple assigment 
`<var1>, <var2> = <value1>, <value2>`

- The left side of the assigment can be  a tuple/list of variables
- The right side of the assigmnet can be any iterable.
    - The iterable is unpacked into the variables
    - The number of itemns of the iterables must match the number of variables
-**Nested  Sequence Unpacking**:
    - <var1>, (<var>, <var3>, ...) = <value1>, (<value2>, <value3>,...),...

````python
a, b, c = 1, 2, 3
print(a, b, c)

a,b,c = (1,2,3)
print(a,b,c)

a,b,c = [1,2,3]
print(a,b,c)

a,b,c = range(1,4)
print(a,b,c)

a,b,c = "123"
print(a,b,c)

(a,b,c) = 1,2,3
print(a,b,c)

[a,b,c] = 1,2,3
print(a,b,c)


````

##### Enumerate

enumerate(iterable, start = 0)

- Creates an iterable sequence representing an enumeration
    - Yields pairs cointaining a count/index and a value yielded by the iterable.
- It is exhausted when the iterable is exhausted
- Lazy Iterable: itemns are created on demand 


````python
Z = ["Blue", "Red", "Yellow", "Green"]
print(enumerate(Z))

for pair in enumerate(Z):
  print(pair)

for i,x in enumerate(Z):
    print(f"indice {i}, valor {x}")
````

````python
Z = [3,4,1,6,2,5]


i_max = 0
x_max = 0
for i, x in enumerate(Z):

  if x > x_max:
    i_max = i

    x_max = x

print(f"El valor mas alto es {x_max} y esta en el indice {i_max}")
````

#### zip(*iterables, strict = False)

- Creates an iterable sequence containing pairs, trios, quartets
    - Yields n-lenght tuples
    - The ith element in every tuple comes from the i-yh iterable
- It is exahusted when the shortest iterable is exahusted
- Lazy iterable, itemns are created on demand (when iterated)



````python
from typing import List
Z1 = ['toyota', 'mercedes', 'mazda', 'audi']
Z2 = ['celica', 'cla', 'rx8', 'a3']

print(zip(Z1,Z2))
print(list(zip(Z1,Z2)))
print(*zip(Z1,Z2))

for pair in zip(Z1,Z2):
  print(pair)

print("-"*20)

for marca, modelo in zip(Z1,Z2):
  print(f'La marca {marca} tiene el modelo {modelo}')
````

````python
names = ["Hernan", "Juan", "Pedro"]

numeros = [123, 234, 567]

searched_name = input('De quien quieres el numero')

for nombre, numero in zip(names, numeros):
  if nombre == searched_name:
    print(numero)
    break
````

## GENERATORS

- Python objects that represents sequences.
    - Many propierties in common with objects like reversed, enumerate or zip.

- Lazy iterables: itemns are created on demand (when iterated) --> Memory Efficiency 
- Single use
    - Once u itearte through a generator and it runs out of values, it becomes permanently empty.
    - Cannot reset, rewind or iterate through it a second time
- Unsized, even infinite
    - sequence of natural numbers, prime numbers
- Can be creating using:
    - Generator Expressions
    - Generator Functions

#### Generator Expressions
- Jus any expresion inside a list Comprehension.
    - [generator expr] --> List Comprehension
    - (generator expr) --> Generator

````python
z = [i*2 for i in range(5)]
g = (i*2 for i in range(5))
print(f"{type(z)=}, {z = }")

print(f"{type(g)=}, {g = }")
print('-'*20)

print(*z)
print(*g)
````

Note that:
- sum([i*i for i in range(1, n+1)])
    - first creates the list, then does the sum.
    - if you increase n, you will exahust ypur system memory (dont do it!)

- sum(i**i for i in range(1, n+1))
    - doesnt create any list, values are created on demand when sum iterates through the generator.
    - U can increase n, the only effect is that u ll wait longer until it finishes.


````python
n = 1_000_000 
print(sum([i ** 2 for i in range(1, n+1)]))

print(sum(i*i for i in range(1, n+1)))
````

#### Generator Functions

- Define sequences by means of the **yield** statement.



````python
def numbers_from_1_to_10():
    for i in range(1, 11):
        yield(
            i
        )

numbers_from_1_to_10()

for i in numbers_from_1_to_10():
    print(i)

print(*numbers_from_1_to_10())
````

****

````python
def first(it, n=10):
  for x in it:
    yield(x)
    n -=1
    if n<=0:
      break
texto = 'This string is to large and contains many characters'
print(*first(texto, n=2), sep="")
print(*first(texto, n=20), sep="")
print(*first(texto, n=50), sep="")
````

**infinite generators**

````python
def naturals():
    i=1
    while True:
        yield i
        i += 1

print(*naturals())
````

#### Generator functions - yield from

````python
def concatenar(it1, it2):
  yield from it1
  yield from it2

print(*concatenar([1,2,3], [4,5,6]))
````

*Example*

````python
def plusminus():
    pm=1
    while True:
        yield pm
        pm *=-1

def row_0_minors(A):
    n = len(A)
    #print(n)
    A = A[1:]
    #print(A)
    for i in range(n):
        yield [[r[j] for j in range(n) if j != i] for r in A]
def det(A):
    if len(A) ==1 :
        return A [0][0]
    else :
        factors = zip(plusminus(), A[0], row_0_minors(A))
        return sum(sign*a*det(minor) for sign,a,minor in factors )
M = [
    [1,2,3],
    [4,5,6],
    [7,8,0]
]
print(*first(plusminus()))
minors = list(row_0_minors(M))
print(minors)
print(det(M))
````

## Reading Text files - open & close

- Python can deal with all kind of files.
    - Packages for managing specific files types: XML, CSV, JSON, etc
- Focus on plain text reading.
- **open(path, mode="r", encoding = None)**
    - Returns a file object that allows text read operations
    - encoding('ascii', 'utf-8'): how decode the text file
- Some properties of a file object:
    - Iterable: The sequence of text lines (a sequence of strings)
        - Lines end with a newline \n character (removable with line.strip())
    - Lazy Iterasble: Lines are read on demand (When files are iterated)
    - file.readline(): returns the next line or empty swtring "" (if file is exhausted)
        - Usde for line in file: If you want to proccess every line
    - file.close(): flush the butter and release teh source

````python
alphabet="abcdefghijklmnopqrstuvwxyz "
f = open("bible.txt", mode="r", encoding="utf8")
# test a few lines
for line in first(f,7):
    
    line = line.rstrip()
    print(f"'{line}'")
    line = "".join((c if c in alphabet else " ") for c in line.lower())
    print(f"--> '{line}'")
f.close()
````

````python
alphabet="abcdefghijklmnopqrstuvwxyz "
f = open("bible.txt", mode="r", encoding="utf8")
s = set()
for line in f:
    # `\n` will be removed because it is not in the alphabet
    #line = line.rstrip()
    line = "".join((c if c in alphabet else " ") for c in line.lower())
    s.update(line.split())
f.close()
print(f"There are {len(s)} different words in the bible")

````

**Reading text files** - with
- Properly closing an open file can be tricky task
    - Even if an exception occurs, we should always try to close it.
- with: a context manager that simplifies resource management.
    - Handles setup and clean up tasks
    - Ensures that allocated resources are properly released.

````python
alphabet="abcdefghijklmnopqrstuvwxyz "
s = set()
with open("bible.txt", mode = "r", encoding="utf-8") as f:
    for line in f:
        line = "".join(
            (
                c if c in alphabet else " "
            ) for c in line.lower()
            )
        print(line)
        s.update(line.split())
print(f"Hay {len(s)} palabras diferentes en la biblia")
````

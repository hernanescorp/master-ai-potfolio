## 2.2 Booleans -- ```bool```

. True and False values.

. Result of relational operators: <, >, <=, >=, !=, ==

. Operands and boolean operators: and, or, not

. Multiple relations: a<b<c same as a<b, b<c.

````python
a = True and False
print(a)

b = True or True
print(b)

c = not True
print(c)
````

````python
a = 3 > 4
print(a)

b = 6 != 9
print(b)

c = 7 == 7
print(c)
````

Boolean casting: ```bool(Something)```

    Numbers
    bool(<zero value>) == False
    bool(< non-zero value>) == True

    Sequences or collections
    bool(<empty>) == False
    bool(<non-empty>) == True

    Conditional Instructions (if, elif, while), bool(), implicity applied:
    



````python
for x in [1, '', " ", [], [1,2,3], range(0), range(2)]:
    if x:
        print(x, ' is:', bool(x))
    else:
        print(x, ' is', bool(x))
````

**Strings-Indexing & Slice Notation**

````python
s = "0123456789"

print(s[-1])
print(s[0])
print(s[5:8]) #includes 5th character
print(s[:9]) #9 not included
print(s[-8:])
print(s[::2])
print(s[2:])
print(s[::-1])
````

**Strings - f-Strings**

````python
from random import randint
a = 1
b = 10

print(f'The number {randint(a,b)} is a random number between {a} and {b}')
````

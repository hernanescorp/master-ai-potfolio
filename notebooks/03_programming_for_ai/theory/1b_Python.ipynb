#### TUPLES

- An immutable sequence of values of any type

- **tuple(it)**: Creates a tuple with the itemns of an iterable object it.
- **Empty tuple**: `()` or `tuple()`

- tuples are indexable, just like lists, and slice compatible
- tuple operators:
    - t1 + t2
    - t * n
    - x in t (and x not in t): True if any of the itemns of t are equal to x

````python
Z = (3, 1.8, -15, "Peter", ['a', 'b', 'c'], 1)

t = tuple(range(5))
print('La tupla t es:', t)

t2 = () #empty tuple
print(bool(t2))

t3 = tuple()
print(bool(t3))

t4 = (4,)
print(type(t4))


print(len(t))

print(t[3])

print(t + t4)

print(t * 3)

print(1 in t)
print(9 in t)



````

#### Secuence Unpacking
- Python Allows multiple assigment 
`<var1>, <var2> = <value1>, <value2>`

- The left side of the assigment can be  a tuple/list of variables
- The right side of the assigmnet can be any iterable.
    - The iterable is unpacked into the variables
    - The number of itemns of the iterables must match the number of variables
-**Nested  Sequence Unpacking**:
    - <var1>, (<var>, <var3>, ...) = <value1>, (<value2>, <value3>,...),...

````python
a, b, c = 1, 2, 3
print(a, b, c)

a,b,c = (1,2,3)
print(a,b,c)

a,b,c = [1,2,3]
print(a,b,c)

a,b,c = range(1,4)
print(a,b,c)

a,b,c = "123"
print(a,b,c)

(a,b,c) = 1,2,3
print(a,b,c)

[a,b,c] = 1,2,3
print(a,b,c)


````

A neural network = many small logistic regressions working together in several layers to learn complex patterns.

### Connectionist model built from complex interactions of simple units

- **Basic Model**: Multi-layer perceptron (MLP). Dense connections of units across multiple layers:
    - **Neurons**: Basic units that receive input, process it and pass the output.
    - **Input layers**: Where data enters.
    - **Hidden Layers**: Intermediate layers where computation happenns.
    - **Output Layer**: Final layer that outputs results.
-**Activation Function**: Determinates the output of neurons; common functions like sigmoid, Tanh, ReLu.

# Why are CNNs so popular?

A CNN (Convolutional Neural Network) is a neural network specialized in images.

Instead of analyzing the whole image at once like a normal MLP, a CNN looks at small regions and detects patterns such as:

- edges
- shapes
- textures

## Main ideas

### Convolution
Small filters move across the image searching for patterns.

### Weight Sharing
The same filter is reused across the entire image, reducing the number of parameters.

### Pooling
The image size is reduced while keeping the most important information.

## Why was it revolutionary?

CNNs:
- need fewer parameters than traditional neural networks
- are more efficient
- learn visual features automatically
- achieved huge improvements in image classification

## Simple pipeline

Image → Convolution → Pooling → Dense layers → Prediction

Example:
- detects edges
- then shapes
- then objects
- finally predicts: “cat”, “dog”, “car”, etc.

## Advantages:
- **Hardware friendly:** Regular structures, well suited for optimization (tensor based computations, parallelization)
- **Scalability**: Hugu models on huge datasets (trillions of parameters over TBs of data).
- **Manage unstructured data**: Can extract from audio and images

## Disavantages:
- **Data hungry**: Their benefits are seen only on extremely Large datasets.
- **Black Boxes**: Predictions cannot be explained or interpreted. Pontential unexpected behaviors. No guarantees.
- **Poor handling of discontinuities**: Smoothness is assumed in the feature space and output. Difficulties with discrete problems.

### 1. Tensor operations following the NumPy API

Frameworks use tensors with operations very similar to NumPy.

A tensor can be:
- a number
- a vector
- a matrix
- or multidimensional data

Example in NumPy:

```python
import numpy as np

x = np.array([1,2,3])
y = x * 2
```

AI frameworks work similarly, but optimized for GPUs and large-scale computation.

Example:

```python
x = tensor([1,2,3])
y = x * 2
```

The goal is:
> perform mathematical operations efficiently using tensor-based computation.

---

### 2. Automatic differentiation for gradient-based optimization

Neural networks learn by adjusting weights to reduce error.

To know how to update those weights, the framework computes derivatives (gradients).

Modern frameworks calculate these gradients automatically using backpropagation.

Example:

```python
loss.backward()
```

The framework automatically:
- computes derivatives
- propagates errors
- updates weights during training

---

### Simple Summary

Modern AI frameworks are mainly built around:

- fast tensor operations similar to NumPy
- automatic gradient computation for training neural networks

````python
import os
os.environ["KERAS_BACKEND"] = "jax"

import keras
import jax

print(jax.devices())

import keras.backend as K
print(K.backend())
````

````python
iris = load_iris()
X = iris.data
y = iris.target
print(iris.DESCR)
print(type(iris))
````

Exploring Data

````python
print(iris.keys()) #Iris is a dictionary

print("features names: \n")
print(iris.feature_names)


print("Iris Data: \n")
print(iris.data[:5])

print("Target names: \n")
print(iris.target_names)

print("Target: \n")
print(iris.target)
````

````python
# Splitting data to train/test


x_train, x_test, y_train, y_test = train_test_split(X,y, test_size = 0.3, random_state = 42)
print("x_train before scaler: \n",x_train[:5])
print("\n")
print("x_test before scaler: \n",x_test[:5])

scaler = StandardScaler() # Normalize the data to give the same scale to all the features
x_train = scaler.fit_transform(x_train) #Use the scaler to normalize the data, fit learns the mean and std of the data and transform the data
x_test = scaler.transform(x_test)
print("\n")
print("x_train after scaler: \n",x_train[:5])
print("\n")
print("x_test after scaler: \n",x_test[:5])

````

````python
#CREATE THE MODEL
model1 = keras.models.Sequential( #Secuential model
    [
        keras.layers.Input(shape=(4,)), #Each data has 4 features

        #First hidden layer with 2 neurons, relu let the positive values and set the negative to zero
        keras.layers.Dense(2, activation = 'relu'),

        #Second hidden layer with 6 neurons, increase the number of neurons to increase the complexity
        keras.layers.Dense(6, activation = 'relu'),

        #Third layer with 3 neurons, softmax is used for multiclass classification, it will return a probability for each class
        keras.layers.Dense(3, activation = 'softmax')
    ]
)

# Compile the way the model will be trained
model1.compile(optimizer='adam', #algorithm to minimize the loss function
               loss = 'sparse_categorical_crossentropy', #error function to classification when labels are integers
               metrics = ['accuracy']) # metric to evaluate the % of correct predictions

# Print the architecture of the model
model1.summary()
````

````python
# Fit the model and see the trainig behavior
#fit means adjust the weights to minimize the loss function
# epoch means how many times the model will see the data
# batch_size means how many data will be used to adjust the weights each epoch 
history1 = model1.fit(x_train, y_train, epochs = 50, batch_size = 5, validation_split = 0.2)
````

````python
# Loss is the error function, the lower the better
# val_loss is the error function on the validation set, the lower the better
# val_accuracy is the accuracy on the validation set

plt.plot(history1.history['loss'], label = 'Loss')
plt.plot(history1.history['val_loss'], label = 'Validation Loss')
plt.legend()
plt.show()
````

````python
#evaluate the model with data that the model has not seen
y_pred_probs = model1.predict(x_test)
print("y_pred_probs: \n", y_pred_probs[:5])
print("\n")
# y_pred is the predicted class
y_pred = np.argmax(y_pred_probs,axis=1)
print("y_pred: \n", y_pred[:5])
print("\n")

print("Accurancy on test set:", accuracy_score(y_test, y_pred))
print("Classification report: \n", classification_report(y_test, y_pred))

conf_matrix = confusion_matrix(y_test, y_pred)
print("Confusion Matrix: \n", conf_matrix)
````

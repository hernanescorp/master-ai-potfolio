## Main idea

A simple logistic regression:

- learns simple patterns,
- linear separations,
- basic classification problems.

A neural network:

- combines MANY neurons,
- organizes neurons into layers,
- learns complex, non-linear patterns.

---

## Typical architecture

```text
Input Layer
    ↓
Hidden Layers
    ↓
Output Layer
```

Each layer learns more abstract representations.

Example in computer vision:

```text
edges → shapes → objects → faces
```

---

## Why are they so powerful?

Because they can learn:

- images
- voice
- text
- natural language
- complex predictions
- generative models

---

## Final idea

```text
A neural network = many small logistic regressions working together in several layers to learn complex patterns.
```

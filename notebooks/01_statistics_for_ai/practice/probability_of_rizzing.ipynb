## Vision Frecuentista

```python
def rizz_frecuentist(true_p = total_smash/total_filas, intentos = 10):
  aproximaciones = np.random.binomial(1, true_p, intentos)
  smashes = np.sum(aproximaciones)

  prob_frec = smashes / intentos

  # Modelizamos que nos den el ig 3 veces seguidas:
  true_pass_ig = true_p ** 3
  estimated_pass_ig = prob_frec ** 3

  fig, ax = plt.subplots(figsize = (10,6))

  # Representamos la funcion del modelo de 3 instas seguidos
  p_range = np.linspace(0,1,100)
  pass_values = p_range **3
  ax.plot(p_range, pass_values, label = 'P(insta 3 veces)')

  ax.scatter(true_p, true_pass_ig, color = 'black', label = 'P(ig real)')
  ax.scatter(prob_frec, estimated_pass_ig, color = 'red', marker ='x', label = 'P(ig estimada)')

  ax.set_xlabel('Rizz Bias(p)')
  ax.set_ylabel('P(3*ig)')
  ax.legend()
  plt.show()


  print('Detalle intentos: ',aproximaciones,' Babies que te dan el ig: ', smashes)
```

```python
rizz_frecuentist(intentos=10)
```

## Bayesian Way

```python
# Total de filas
total_filas = len(df)

# Total de Smash
total_smash = (df["score"] == "Smash").sum()

print("Total filas:", total_filas)
print("Total Smash:", total_smash)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta, binom
```

```python
from matplotlib.lines import lineStyles
def bayesian_rizz(smashes = total_smash,intentos = total_filas, alpha_prior = 2, beta_prior = 10):
  #Frecuentist
  p_mle = smashes / intentos

  #Bayesian (MAP estimate with Beta Prior)
  p_map_prior = (alpha_prior - 1) / (alpha_prior + beta_prior - 2)

  #Posterior Map
  p_map_posterior = (smashes + alpha_prior -1) / (intentos + alpha_prior + beta_prior - 2)

  print(f"Dieron Insta: {smashes}, rechazos {intentos - smashes}")
  print(f"p frecuentista de rizz {p_mle}")
  print(f"p MAP prior de rizz {p_map_prior}")
  print(f"p MAP posterior de rizz {p_map_posterior}")


  p_axis = np.linspace(0,1,100)
  plt.title('Posterior Rizzin Probability')
  plt.plot(p_axis, p_axis ** 3, color='black', linestyle= '--', label='Probailidad 3 veces ig P=p^3')

  prior_pdf = beta.pdf(p_axis, alpha_prior, beta_prior)
  plt.fill_between(p_axis, 0 , prior_pdf / prior_pdf.max(), alpha = 0.3, label = 'Prior Beta(2,10)')

  posterior_pdf = beta.pdf(p_axis, alpha_prior + smashes, beta_prior + intentos - smashes)
  plt.fill_between(p_axis, 0, posterior_pdf / posterior_pdf.max(), color = 'green', label='Bayesian Posteriro', alpha = 0.2)

  plt.scatter(p_mle, p_mle**3, label = 'MLE estimated')
  plt.scatter(p_map_prior, p_map_prior ** 3, label = 'MAP Prior', color = 'blue' )
  plt.scatter(p_map_posterior, p_map_posterior ** 3, label = 'MAP Posterior')

  plt.legend()
  plt.show()

```

```python
bayesian_rizz()
```

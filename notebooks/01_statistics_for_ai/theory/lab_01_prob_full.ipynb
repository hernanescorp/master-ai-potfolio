## UNIVARIATE DATA

### GAUSSIAN DISTRIBUTION

```python
#Define a Gaussian distribution
#A Gaussian distribution is a continuous probability distribution that is symmetric about the mean, showing that data near the mean are more frequent in occurrence than data far from the mean.
#The Gaussian distribution is also known as the normal distribution.
mean = 0

#the std is the standard deviation, which is how much the data is spread out
std1 = 1
std2 = 2
std3 = 0.5


true_dist = stats.norm(mean, std1)
true_dist2 = stats.norm(mean, std2)
true_dist3 = stats.norm(mean, std3)
```

Probability density function and Cumuative Distribution Function

```python
x = np.linspace(-5,5,1000)
plt.plot(x, true_dist.pdf(x), label=f'std = {std1}')
plt.plot(x, true_dist2.pdf(x), label = f'std ={std2}')
plt.plot(x, true_dist3.pdf(x), label = f'std = {std3}')
plt.legend()
plt.grid(True)
```

```python
    plt.plot(x, true_dist.cdf(x), label='std = 1')
    plt.plot(x, true_dist2.cdf(x), label = 'std = 2')
    plt.plot(x, true_dist3.cdf(x), label = 'std = 0.5')
    plt.title('CDF')
    plt.legend()
    plt.grid(True)
```

## Drawing or Sampling From a Gaussian distribution

```python
## Draw samples from the gaussian distribution
n_samples = 1000
# rvs method of scipy.stats.norm instance
# Draw random samples from a normal (Gaussian)
samples = true_dist.rvs(n_samples)
print(samples[:10])
```

### Sufficient statistics of observed samples




La media empírica es:

$$
\bar{x} = \frac{1}{n}\sum_{i=1}^{n} x_i
$$

La varianza empírica es:

$$
\sigma^2 = \frac{1}{n}\sum_{i=1}^{n}(x_i - \bar{x})^2
$$

La desviación estándar empírica es:

$$
\sigma = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(x_i - \bar{x})^2}
$$

```python
empirical_mean = np.sum(samples)/n_samples
empirical_std = np.sqrt(np.sum((samples-empirical_mean)**2)/n_samples)
empirical_var = np.sum((samples - empirical_mean)**2)/n_samples
```

```python
sample_mean = np.mean(samples)
sample_std = np.std(samples)
sample_var = np.var(samples)
```

```python
#Mean comparison:
print('True mean:', mean)
print('Empirical mean: ', empirical_mean)
print('Sample mean: ', sample_mean)
print('Empirical mean to true mean MSE:', np.mean((empirical_mean - mean)**2))
print('Sample mean to true mean: MSE: ', np.mean((sample_mean - mean)**2))
```

```python
# Variance Comparison
print('True variance:', std1 ** 2)
print('Empirical variance: ', empirical_var)
print('Sample variance: ', sample_var)
print('Empirical variance to true variance MSE: ', np.mean((empirical_var - std1 **2)**2))
print('Sample variance to true variance MSE: ', np.mean((sample_var - std1 **2)**2))
```

### Cumulative probability of samples: CDF 

The cumulative distribution function of a real valued random variable X is the function given by:

$$ F_{X}(x)=\operatorname {P} (X\leq x) $$

where the right hand side represents the probability that the random variable X takes on a value less than or equal to x.

```python
def compute_empirical_cdf(x, samples):
    n_samples = len(samples)
    emp_cdf = np.zeros_like(x)
    for i, x_val in enumerate(x):
        emp_cdf[i] = np.sum(samples <= x_val) / n_samples
    return emp_cdf
```

```python
print('Axis X with linespace:', x[0:10])
print('Samples: ',samples[0:10])

emp_cdf = compute_empirical_cdf(x, samples)
print(emp_cdf[0:10])


plt.plot(
    x,
    emp_cdf,
    label = 'Empirical CDF'
    )
plt.plot(
    x,
    true_dist.cdf(x),
    label = 'True cdf'
)
plt.legend()
plt.show()
```

```python
# Compute the probability using the empirical cdf
emp_prob_event_cdf = emp_cdf[np.argmax(x>=b)] - emp_cdf[np.argmax(x>=a)]

# Compute teh probability of the event using the tru cdf

true_prob_event = true_dist.cdf(b) - true_dist.cdf(a)
```

```python
#Print the results and compare
print('Empirical probability of the event:', emp_prob_event)
print('Empirical probability of the event cdf:', emp_prob_event_cdf)
print('True probability:', true_prob_event)

print('Empirical probability to the true probability:', np.mean((emp_prob_event - true_prob_event)**2))

print('Empirical probability to the true probability MSE cdf: ', np.mean((emp_prob_event_cdf - true_prob_event)**2))
```

### Study how the computation the probability of the event changues, over different number of samples

```python
a = -1
b = 1

n_samples_list = np.arange(10,10000, 100) # list from 10 to 10000 with 100 elements
emp_prob_event_list = []

for n_samples in n_samples_list:
    samples = true_dist.rvs(n_samples)
    event = (samples >=a) & (samples <=b)
    emp_prob_event = np.sum(event) / n_samples
    emp_prob_event_list.append(emp_prob_event)

print('Empirical PROBABILITIES:',len(emp_prob_event_list))

true_prod_event = true_dist.cdf(b) - true_dist.cdf(a)

plt.plot(
    n_samples_list,
    emp_prob_event_list,
    label = 'Empirical Probability Event'
)
plt.plot(
    n_samples_list,
    [true_prob_event] * len(n_samples_list),
    label = 'True Probability Event'
)

plt.legend()
plt.show()
```

### Study how with more data, the empirical distributions converge to the true distribution

```python
plt.plot(
    x,
    true_dist.pdf(x),
    label = 'True Distribution',
    linewidth = 3
)

for n_samples in [10, 100, 1000, 10000]:
    samples = true_dist.rvs(n_samples)
    emp_dist = stats.gaussian_kde(samples)
    plt.plot(
        x, 
        emp_dist.pdf(x),
        label = f'Emprirical distribution {n_samples}'
    )
plt.legend(loc="upper left")
plt.show()
```
